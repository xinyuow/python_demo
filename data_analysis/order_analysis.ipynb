# 模拟电商订单数据分析


> 需求：创建5行5列1-100的随机数组，完成【切片、修改、添加、删除、查找】全流程

```python
import numpy as np
```

```python
# 创建一个5行5列1-100的随机数组
np.random.seed(100)
arr = np.random.randint(1, 101, size=(5, 5))
print(f"原始订单数据:\n {arr}")
```

```python
# 获取前三个订单的前三项数据
arr_slice = arr[:3, :3].copy()
print(f"前三个订单的前三项数据:\n {arr_slice}")
```

```python
# 修改切片中，大于50的元素，将其设为50
arr_slice[arr_slice > 50] = 50
print(f"修改后的前三个订单的前三项数据:\n {arr_slice}")
```

```python
# 新增一笔订单数据
arr_new = np.append(arr, [[100]*5], axis=0)
print(f"新增一笔订单数据后的:\n {arr_new}")
```

```python
# 删除第5行订单数据
arr_new = np.delete(arr_new, 5, axis=0)
print(f"删除第三行订单数据后的:\n {arr_new}")
```

```python
# 查找所有大于50的订单
row_id, col_id = np.where(arr_new > 50)
print(f"所有大于50的订单的行索引: {row_id}")
print(f"所有大于50的订单的列索引: {col_id}")
print(f"所有大于50的订单的元素值: {arr_new[row_id, col_id]}")
```
